```python
def program_pembelian_bensin():
    """Program pembelian bensin sederhana."""

    harga_per_liter = 15000  # Harga bensin per liter (contoh)
    jenis_bensin = ["Pertalite", "Pertamax", "Pertamax Turbo", "shell super"]  # Daftar jenis bensin

    print("Selamat datang di SPBU!")
    print("Jenis Bensin yang Tersedia:")
    for i, bensin in enumerate(jenis_bensin):
        # Display options starting from 1 to match list indexing
        print(f"{i+1}. {bensin}")

    while True:
        try:
          pilihan_bensin = int(input("Pilih jenis bensin (nomor): "))
          # Adjust the condition to match the new display numbering
          if 1 <= pilihan_bensin <= len(jenis_bensin):
              break
          else:
              print("Pilihan tidak valid. Masukkan nomor yang sesuai.")
        except ValueError:
          print("Masukkan angka yang valid")

    while True:
        try:
            jumlah_liter = float(input("Masukkan jumlah liter bensin: "))
            if jumlah_liter > 0:
                break
            else:
                print("Jumlah liter harus lebih dari 0.")
        except ValueError:
            print("Masukkan angka yang valid.")

    total_harga = jumlah_liter * harga_per_liter
    # Access the list element directly using pilihan_bensin - 1
    print(f"\nJenis Bensin: {jenis_bensin[pilihan_bensin-1]}")
    print(f"Jumlah Liter: {jumlah_liter} liter")
    print(f"Harga per liter: Rp {harga_per_liter}")
    print(f"Total Harga: Rp {total_harga}")

    # Tambahkan input pembayaran dan kembalian jika diperlukan

if __name__ == "__main__":
    program_pembelian_bensin()
```

```python
from scipy.interpolate import interp1d
import matplotlib.pyplot as plt
```

```python
x = [0, 1, 2]
y = [1, 3, 2]

f = interp1d(x,y)

y_hat = f(1.5)
print(y_hat)
```

```python
plt.figure(figsize=(10, 8)) # Ukuran plot 10x8 inci
plt.plot(x, y, '-ob') # Menampilkan titik data dengan garis biru dan lingkaran
plt.plot(1.5, y_hat, 'ro') # Menampilkan titik haris interpolasi dengan garis merah dan lingkaran
plt.title('Lienar interpolation at x = 1.5') # Judul grarik
plt.xlabel('x') # Nama label sumbu x
plt.ylabel('y') # Nama label sumbu y
plt.show() # Menampilkan grafik

```

```python
import numpy as np
import matplotlib.pyplot as plt
def divided_diff(x,y):
  n = len(y)
  coef = np.zeros([n, n]) # Membuat tabel kosong untuk divided differences
  coef[:,0] = y # Kolom pertama berisi nilai y

  for j in range(1,n): # Loop untuk mengisi kolom - kolom beriktutnya
    for i in range(n-j):
      coef[i][j] = (coef[i+1][j-1] - coef[i][j-1]) / (x[i + j] - x[i])

  return coef
def newton_poly(coef, x_data, x):
      n = len(x_data) - 1
      p = coef[n]

      for k in range(1,n+1):
        p = coef[n-k] + (x - x_data[n-k])*p

      return p
```

```python
x = np.array([-5, -1, 0, 2])
y = np.array([-2, 6, 1, 3])

# Menghitung koefisien divided difference
a_s = divided_diff(x,y)[0,:]

# Evaluasi interpolasi pada titik - titik baru
x_new = np.arange(-5, 2.1, .1)
y_new = newton_poly(a_s, x, x_new)

# Membuat plot
plt.figure(figsize=(12, 8))
plt.plot(x, y, 'bo') # Membuat titik data asli
plt.plot(x_new, y_new, 'g') # Membuat kurva interpolasi

```

```python
from scipy.interpolate import CubicSpline
import numpy as np
import matplotlib.pyplot as plt

x = [0, 1, 2] # Titik data pada sumbu x
y = [1, 3, 2] # Titik data pada sumbu y

# Membuat fungsi interpilasi kubik dengan batas 'natural'
f = CubicSpline(x, y, bc_type='natural')

# Membuat rentang nilai x untuk plot hasil interpolasi
x_new = np.linspace(0, 2, 100) # 100 titik antara 0 dan
y_new = f(x_new) # Menghitung nilai y mengunakan interpolasi kubik

# membuat plot
plt.figure(figsize=(10, 8)) # Ukuran plot 10x8 inci
plt.plot(x_new, y_new, 'g') # Menampilkan kurva interpolasi dalam warna biru
plt.plot(x, y, 'ro') # menampilkan titik data asli dalam warna merah
plt.title('Cubic Spline Interpolation') # Judul grafik
plt.xlabel('x') # Nama label sumbu x
plt.ylabel('y') # Nama label sumbu y
plt.show() # Menampilkan grafik
```

```python
from scipy.interpolate import interp1d
import numpy as np

# Modul lab

# Data jarak lari
x = np.array([0, 5, 10])
y = np.array([10, 15, 20])

f = interp1d(x, y, kind='linear')

# Estimate speed at 7.2 meters
speed_at_7_2m = f(7.2)

y_hat = f(7.2)
print(f"Kecepatan lari Ben Johnson pada jarak 7.2 meter: {speed_at_7_2m} m/detik")
```

```python
plt.figure(figsize=(10, 8)) # Ukuran plot 10x8 inci
plt.plot(x, y, '-ob') # Menampilkan titik data dengan garis biru dan lingkaran
plt.plot(7.2, y_hat, 'ro') # Menampilkan titik haris interpolasi dengan garis merah dan lingkaran
plt.title('Lienar interpolation at x = 7.2') # Judul grarik
plt.xlabel('x') # Nama label sumbu x
plt.ylabel('y') # Nama label sumbu y
plt.show() # Menampilkan grafik
```
