```python
import numpy as np
def my_bisection (f, a, b, tol):
  #approximate a root, R. of bounded
  #by a and b to within tolerance
  #|f(m)|<tol with athe midpoint
  #bettwen a and b recursive implementation



  #check if a and b bound a root
  if np.sign(f(a)) == np.sign(f(b)):
    raise Exception(
        "the scalars a and b do not bound a root")

  #get midpoint
  m = (a+b)/2

  if np.abs(f(m))<tol:
    #stopping condition, report m as root
    return m
  elif np.sign(f(a))==np.sign(f(m)):
    #case where m is an improvement on a.
    #make recursive call with a=m
    return my_bisection(f, m, b, tol)
  elif np.sign(f(b))==np.sign(f(m)):
    #case where m is an improvement on b.
    #make recursive call with b=m
    return my_bisection(f, a, m, tol)
```

```python
f=lambda x: x**2-2
r1=my_bisection(f, 0, 2, 0.1)
print("r1=", r1)
r01=my_bisection(f, 0, 2, 0.01)
print("r01=", r01)

print("f(r1)=", f(r1))
print("f(r0)", f(r01))
```

```python
def my_newton(f, df, x0, tol):
  #output is an estimationof the root of f
  #using the newton raphson method
  #recursive implementation
    if abs(f(x0)) < tol:
        return x0
    else:
        return my_newton(f, df, x0 - f(x0)/df(x0), tol)
```

```python
import math

# Bisection Method
def f_bisection(x):
    return math.sin(x) + x

def bisection(a, b, tol):
    if f_bisection(a) * f_bisection(b) >= 0:
        print("Bisection method gagal. f(a) dan f(b) harus beda tanda.")
        return None

    iterasi = 0
    while (b - a) / 2 > tol:
        iterasi += 1
        c = (a + b) / 2
        if f_bisection(c) == 0:
            break
        elif f_bisection(a) * f_bisection(c) < 0:
            b = c
        else:
            a = c
        print(f"Iterasi {iterasi}: c = {c:.5f}")

    return (a + b) / 2
```

```python
import math

def f(x):
    return math.sin(x) + x

def bisection(a, b, tol):
    while (b - a) / 2 > tol:
        c = (a + b) / 2
        if f(c) == 0:
            return c
        elif f(a) * f(c) < 0:
            b = c
        else:
            a = c
    return (a + b) / 2

# Panggil fungsi
akar = bisection(0, 1, 0.01)
print("Akar hampiran (Bisection):", akar)
```

Newton-Raphson Method


```python
def f(x):
    return x*3 - 3*x*2 + 2*x

def df(x):
    return 3*x**2 - 6*x + 2

def newton_raphson(x0, tol):
    while True:
        x1 = x0 - f(x0) / df(x0)
        if abs(x1 - x0) < tol:
            break
        x0 = x1
    return x1

# Panggil fungsi
akar = newton_raphson(1.5, 0.01)
print("Akar hampiran (Newton-Raphson):", akar)
```
